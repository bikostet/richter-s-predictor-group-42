# tuto glmnet 

Voir : https://glmnet.stanford.edu/articles/glmnet.html

```r
data(QuickStartExample)
x <- QuickStartExample$x
y <- QuickStartExample$y
```

```r
fit <- glmnet(x, y)
```

```r
plot(fit)
```

```r
print(fit)
```

```r
coef(fit, s = 0.1)
```

```r
set.seed(29)
nx <- matrix(rnorm(5 * 20), 5, 20)
predict(fit, newx = nx, s = c(0.1, 0.05))
```

Pour trouver la meuilleur valeur de lambda 

```r
cvfit <- cv.glmnet(x, y)
```

```r
plot(cvfit)
```

```r
cvfit$lambda.min
```

```r
coef(cvfit, s = "lambda.min")
```

```r
predict(cvfit, newx = x[1:5,], s = "lambda.min")
```

```r
wts <-  c(rep(1,50), rep(2,50))
fit <- glmnet(x, y, alpha = 0.2, weights = wts, nlambda = 200)
```
